```python
#CREACION DE COPIA DE TRABAJO
orders_work = orders.copy()
catalog_work=catalog.copy()
marketing_work=marketing.copy()
```

```python
# REVISION Y CALIDAD DE DATOS DE ORDERS

#Correccion de formato de fecha
orders_work['fecha_hora_pedido'] = pd.to_datetime(orders_work['fecha_hora_pedido'])
orders_work.info()
```

```python
#Conteo de NA por columna
orders_work.isna().sum()
```

```python
#Conteo de NA en total del data set Orders
orders_work.isna().sum().sum()
```

```python
#Revision de columnas que tienen valores ausentes
orders_work.columns[orders_work.isna().any()]
```

```python
# Porcentaje de valores ausentes por columna
porcentaje_na = (orders_work.isna().sum() / len(orders_work)) * 100
porcentaje_na = porcentaje_na.round(2)

# Ordenado de mayor a menor para priorizar
print(porcentaje_na.sort_values(ascending=False))
```

```python
#Se observa que el % de valores ausentes el mayor de 1.20 % y todos los demas son menores, por lo que se toma la decision de va a reemplazar con la moda y la media
```

```python
# Categóricas → moda
orders_work['pais'] = orders_work['pais'].fillna(orders_work['pais'].mode()[0])
orders_work['dispositivo'] = orders_work['dispositivo'].fillna(orders_work['dispositivo'].mode()[0])
orders_work['fuente_referencia'] = orders_work['fuente_referencia'].fillna(orders_work['fuente_referencia'].mode()[0])
orders_work['nombre_producto'] = orders_work['nombre_producto'].fillna(orders_work['nombre_producto'].mode()[0])
orders_work['categoria_producto'] = orders_work['categoria_producto'].fillna(orders_work['categoria_producto'].mode()[0])

# Numéricas → media
orders_work['cantidad'] = orders_work['cantidad'].fillna(orders_work['cantidad'].mean())
orders_work['precio_unitario'] = orders_work['precio_unitario'].fillna(orders_work['precio_unitario'].mean())
orders_work['monto_descuento'] = orders_work['monto_descuento'].fillna(orders_work['monto_descuento'].mean())

print(orders_work.isna().sum())
```

<div class="alert alert-block alert-warning">
<b>Comentario del revisor. (Iteración 1)</b> <a class="tocSkip"></a><br><br>

Realizaste un tratamiento explícito de los valores ausentes, lo cual es positivo. Sin embargo, conviene revisar con un poco más de detalle la imputación de <code>cantidad</code>, <code>nombre_producto</code> y <code>categoria_producto</code>.

En esta sección, la media inicial de <code>cantidad</code> está afectada por valores extremos y posteriormente las filas imputadas pueden terminar eliminándose con el filtro del IQR. Además, completar producto y categoría de manera independiente con sus respectivas modas podría crear combinaciones que no existían originalmente.

En un escenario profesional, estas decisiones deben mantener la coherencia entre producto, categoría, cantidad y monto. Para fortalecer la limpieza, revisa cuántas filas son afectadas por cada imputación y decide si conviene recuperar el dato mediante otra columna relacionada, conservar una categoría como “desconocido” o retirar únicamente los registros que no puedan validarse.

</div>

```python
orders_work.describe()
```

```python
#Revision continua de limpieza de datos por valores que son inconsistentes 

neg_cant = (orders_work['cantidad'] < 0).sum()
cero_cant = (orders_work['cantidad'] == 0).sum()
print(f"cantidad negativa: {neg_cant} | cantidad cero: {cero_cant}")

Q1 = orders_work['cantidad'].quantile(0.25)
Q3 = orders_work['cantidad'].quantile(0.75)
IQR = Q3 - Q1
limite_superior = Q3 + 3 * IQR
print(f"Límite superior IQR para cantidad: {limite_superior}")
```

```python
# Aplicar filtro de cantidad
filas_antes = len(orders_work)

orders_work = orders_work[
    (orders_work['cantidad'] > 0) &
    (orders_work['cantidad'] <= limite_superior)
]

filas_despues = len(orders_work)
print(f"Filas eliminadas por cantidad: {filas_antes - filas_despues}")
print(orders_work['cantidad'].describe())
```

```python
print("Filas con cantidad > 5:", (orders_work['cantidad'] > 5).sum())
print(orders_work[orders_work['cantidad'] > 5]['cantidad'].value_counts().sort_index())
```

```python
# Diagnóstico de negativos
neg_monto = (orders_work['monto_total'] < 0).sum()
print(f"monto_total negativo: {neg_monto}")
print(orders_work[orders_work['monto_total'] < 0][['id_pedido','cantidad','precio_unitario','monto_descuento','monto_total']].head(10))

# Diagnóstico de outliers extremos
p99 = orders_work['monto_total'].quantile(0.99)
p999 = orders_work['monto_total'].quantile(0.999)
print(f"Percentil 99: {p99}")
print(f"Percentil 99.9: {p999}")
print(f"Máximo actual: {orders_work['monto_total'].max()}")

# Revisa cuántas filas quedan por encima de distintos cortes
print("Filas > p99*3:", (orders_work['monto_total'] > p99*3).sum())
print("Filas > 10000:", (orders_work['monto_total'] > 10000).sum())
```

```python
# Consistencia aritmética
orders_work['monto_calculado'] = (
    orders_work['cantidad'] * orders_work['precio_unitario']
    - orders_work['monto_descuento']
)
orders_work['diff_monto'] = (
    orders_work['monto_total'] - orders_work['monto_calculado']
).abs()

inconsistentes = orders_work[orders_work['diff_monto'] > 1]  # tolerancia de redondeo
print(f"Filas con inconsistencia aritmética: {len(inconsistentes)}")
print(inconsistentes[['id_pedido','cantidad','precio_unitario','monto_descuento','monto_total','monto_calculado','diff_monto']].head(10))
print(orders_work['diff_monto'].describe())
```

```python
# Cierre del pipeline de limpieza
orders_work = orders_work.drop(columns=['monto_calculado', 'diff_monto'])

print(f"Dataset final: {orders_work.shape[0]} filas, {orders_work.shape[1]} columnas")
print(orders_work.describe())
```

### 2.1 Cálculo de KPIs principales

**🎯 Objetivo:** Calcular los indicadores clave del negocio para evaluar ingresos, costos y rentabilidad.

Se usan los 3 datasets (`orders`, `catalog`, `marketing_spend`):

**📊 Parte 1: Rentabilidad del negocio**
- ¿Cuál es el ingreso total (revenue)? 
- ¿Cuál es el costo total? 
- ¿Cuánto se ha invertido en marketing? 
- ¿El negocio es rentable? (calcular profit)  

---

**📈 Parte 2: Comportamiento de ventas**
- ¿Cuál es el ticket promedio por orden? 
- ¿Cuál es la cantidad promedio de productos por orden? 
- ¿Cuál es el producto más vendido?
- ¿Cuánto se ha gastado en marketing por canal? 

```python
orders1= pd.read_csv('orders_clean.csv')
catalog1= pd.read_csv('catalog_clean.csv')
marketing1= pd.read_csv('marketing_clean.csv')
```

```python
# Parte 1: Rentabilidad del negocio
```

```python
#1. Ingreso total (Revenue)
revenue_total = orders1['monto_total'].sum()
print(f"Ingreso total: ${revenue_total:,.2f}")
```

```python
# Merge para traer costo_unitario a cada pedido
orders_con_costo = orders1.merge(
    catalog1[['nombre_producto', 'costo_unitario']],
    on='nombre_producto',
    how='left'
)
# Verificar que el merge no generó NaN (productos sin match en el catálogo)
print("Pedidos sin costo_unitario tras el merge:", orders_con_costo['costo_unitario'].isna().sum())

# Costo total = costo_unitario * cantidad, sumado
orders_con_costo['costo_total_linea'] = orders_con_costo['costo_unitario'] * orders_con_costo['cantidad']
costo_total = orders_con_costo['costo_total_linea'].sum()
print(f"Costo total: ${costo_total:,.2f}")
```

```python
# Inversión total en marketing
marketing_total = marketing1['gasto'].sum()
print(f"Inversión total en marketing: ${marketing_total:,.2f}")
```

```python
# Profit y rentabilidad
profit = revenue_total - costo_total - marketing_total
margen_pct = (profit / revenue_total) * 100

print(f"Profit: ${profit:,.2f}")
print(f"Margen de rentabilidad: {margen_pct:.2f}%")
print("¿Es rentable?:", "Sí" if profit > 0 else "No")
```

<div class="alert alert-block alert-success">
<b>Comentario del revisor. (Iteración 1)</b> <a class="tocSkip"></a><br><br>

Muy buen desarrollo de la sección de rentabilidad.

El merge entre órdenes y catálogo se realiza mediante <code>nombre_producto</code>, verificas que no queden pedidos sin costo y calculas el costo total como <code>costo_unitario * cantidad</code>. Posteriormente defines el profit como revenue menos costos y menos marketing, obteniendo aproximadamente <b>$2.94 millones</b> y un margen de <b>30.45%</b>.

Esta fórmula está alineada con la rúbrica y permite concluir correctamente que el negocio es rentable después de considerar tanto el costo de los productos como la inversión de marketing.

</div>

```python
# ======================
# PARTE 1: Totales del funnel
# ======================
query_totals = '''
SELECT 
    nombre_evento,
    COUNT(DISTINCT id_usuario) AS usuarios_unicos
FROM events
GROUP BY nombre_evento
ORDER BY 
    CASE nombre_evento
        WHEN 'first_visit' THEN 1
        WHEN 'select_item' THEN 2
        WHEN 'add_to_cart' THEN 3
        WHEN 'begin_checkout' THEN 4
        WHEN 'add_payment_info' THEN 5
        WHEN 'purchase' THEN 6
        ELSE 99
    END
'''
totals = pd.read_sql(query_totals, con=engine)
totals
```

```python
# ======================
# PARTE 2: Análisis de conversión
# ======================
# Tasa de conversión respecto a la primera etapa
totals['pct_del_total'] = (totals['usuarios_unicos'] / totals['usuarios_unicos'].iloc[0] * 100).round(2)

# Tasa de conversión etapa a etapa (drop-off)
totals['pct_etapa_anterior'] = (totals['usuarios_unicos'] / totals['usuarios_unicos'].shift(1) * 100).round(2)

# Caída de usuarios en número absoluto entre etapas consecutivas
totals['usuarios_perdidos'] = totals['usuarios_unicos'].shift(1) - totals['usuarios_unicos']

print(totals)

# Etapa donde se pierde más usuarios (mayor caída absoluta)
etapa_critica = totals.loc[totals['usuarios_perdidos'].idxmax(), 'nombre_evento']
print(f"\nEtapa con mayor pérdida de usuarios: {etapa_critica}")

# Tasa de conversión final (primera etapa → última etapa)
tasa_final = totals['usuarios_unicos'].iloc[-1] / totals['usuarios_unicos'].iloc[0] * 100
print(f"Tasa de conversión final (first_visit → purchase): {tasa_final:.2f}%")
```

## 🔹 Paso 4: Evaluar si los usuarios regresan (retención por cohortes)

**🎯 Objetivo:** Analizar la retención de usuarios para entender si regresan después de registrarse.

**Tablas**

- `users` 
- `user_activity` 

---
1. Se identifica la cohorte de cada usuario según el **mes de registro**.


2. Se calcula la retención semanal: cuántos usuarios **se mantienen activos** en cada semana desde su registro.
   - `retenido_w1`: usuarios activos en la semana 1  
   - `retenido_w2`: usuarios activos en la semana 2  
   - `retenido_w3`: usuarios activos en la semana 3  


3. Se calcula el porcentaje de retención para cada semana, dividiendo los usuarios retenidos entre los clientes iniciales de la cohorte:  
   - `semana_1`: porcentaje de usuarios retenidos en la semana 1  
   - `semana_2`: porcentaje de usuarios retenidos en la semana 2  
   - `semana_3`: porcentaje de usuarios retenidos en la semana 3  

Se revisa que la columna de fecha esté en formato correcto (`DATE`).  
Se realiza la conversión usando: `CAST(fecha_registro AS DATE)`

```python
# Explorar tabla users
query_users = '''
SELECT 
    id_usuario,
    CAST(fecha_registro AS DATE) AS fecha_registro,
    país,
    dispositivo,
    tipo_plan
FROM users
LIMIT 5;
'''
users = pd.read_sql(query_users, con=engine)
print(users.dtypes)
users.head()
```

```python
users['fecha_registro'] = pd.to_datetime(users['fecha_registro'])
print(users.dtypes)
```

```python
# Explorar tabla user_activity
query_activity = '''
SELECT *
FROM user_activity
LIMIT 5;
'''
user_activity = pd.read_sql(query_activity, con=engine)
user_activity['fecha_actividad']=pd.to_datetime(user_activity['fecha_actividad'])
print(user_activity.dtypes)
user_activity.head()
```

```python
# Retención por cohortes
# ======================
query_retencion = '''
WITH usuarios_cohorte AS (
    SELECT
        id_usuario,
        DATE_TRUNC('month', CAST(fecha_registro AS DATE)) AS cohorte
    FROM users
),
actividad_semanal AS (
    SELECT
        id_usuario,
        MAX(CASE WHEN dias_despues_registro = 7  THEN activo ELSE 0 END) AS retenido_w1,
        MAX(CASE WHEN dias_despues_registro = 14 THEN activo ELSE 0 END) AS retenido_w2,
        MAX(CASE WHEN dias_despues_registro = 21 THEN activo ELSE 0 END) AS retenido_w3
    FROM user_activity
    GROUP BY id_usuario
)
SELECT
    uc.cohorte,
    COUNT(DISTINCT uc.id_usuario) AS clientes_iniciales,
    SUM(COALESCE(a.retenido_w1, 0)) AS retenido_w1,
    SUM(COALESCE(a.retenido_w2, 0)) AS retenido_w2,
    SUM(COALESCE(a.retenido_w3, 0)) AS retenido_w3
FROM usuarios_cohorte uc
LEFT JOIN actividad_semanal a ON uc.id_usuario = a.id_usuario
GROUP BY uc.cohorte
ORDER BY uc.cohorte
'''

retencion = pd.read_sql(query_retencion, con=engine)
retencion
```

```python

#Calculo de porcentaje de retencion
retencion['semana_1'] = (retencion['retenido_w1'] / retencion['clientes_iniciales'] * 100).round(2)
retencion['semana_2'] = (retencion['retenido_w2'] / retencion['clientes_iniciales'] * 100).round(2)
retencion['semana_3'] = (retencion['retenido_w3'] / retencion['clientes_iniciales'] * 100).round(2)

retencion
```

Actualmente utilizas condiciones exactas para los días 7, 14 y 21:

## 🔹 Paso 5: Validar si los cambios generan impacto (test estadístico)

🎯 **Objetivo:** Evaluar si la modificación en la UI del checkout impacta la **tasa de conversión de compra**.

---

1. **Analizar el dataset** `experiment_checkout_ui.csv` para identificar la métrica principal **conversion**.
   - La métrica **conversion** es 1 si el usuario completó la compra, 0 si no.    
2. **Plantear la hipótesis estadística**     
3. **Aplicar el test estadístico adecuado** 
4. **Interpretar el resultado**  

---
Hipótesis estadística
   - **H₀ (Hipótesis nula):** No existe diferencia entre la tasa de conversion del grupo de control y de tratamiento
   - **H₁ (Hipótesis alternativa):** Existe una diferencia en la tasa de conversion entre ambos grupos
   
**Test estadístico:** .z  
**Nivel de significancia alpha:** 0.05

```python
experiment_checkout_ui=pd.read_csv('experiment_checkout_ui.csv')
```

```python
experiment_checkout_ui.info()
```

```python
# Paso 1: Explorar el dataset y confirmar la métrica
print(experiment_checkout_ui.shape)
print(experiment_checkout_ui.dtypes)
experiment_checkout_ui.head()

# Confirmar los grupos experimentales
print(experiment_checkout_ui['variante'].unique())
```

```python
# Aplicacion del test estadistico
from statsmodels.stats.proportion import proportions_ztest

# Tasa de conversión por grupo
resumen = experiment_checkout_ui.groupby('variante')['convirtio'].agg(
    conversiones='sum', usuarios='count', tasa_conversion='mean'
)
print(resumen)

# Preparar datos para el test
control = experiment_checkout_ui[experiment_checkout_ui['variante'] == 'control']['convirtio']
tratamiento = experiment_checkout_ui[experiment_checkout_ui['variante'] == 'tratamiento']['convirtio']

count = [tratamiento.sum(), control.sum()]   # conversiones exitosas por grupo
nobs = [len(tratamiento), len(control)]      # tamaño de cada grupo

z_stat, p_value = proportions_ztest(count, nobs, alternative='two-sided')

print(f"\nZ-statistic: {z_stat:.4f}")
print(f"p-value: {p_value:.4f}")
```

```python
#  Interpretar el resultado
alpha = 0.05
if p_value < alpha:
    print(f"Se rechaza H0 (p={p_value:.4f} < {alpha}): la diferencia en conversión es estadísticamente significativa.")
else:
    print(f"No se rechaza H0 (p={p_value:.4f} >= {alpha}): no hay evidencia suficiente de diferencia en conversión.")
```
